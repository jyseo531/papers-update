```python
import requests
from bs4 import BeautifulSoup

class GitHubStarCrawler:
    def __init__(self, repo_url):
        self.repo_url = repo_url

    def get_star_count(self):
        try:
            response = requests.get(self.repo_url)
            response.raise_for_status()
            soup = BeautifulSoup(response.text, 'html.parser')
            star_tag = soup.find('span', {'class': 'Counter js-social-count'})
            if star_tag:
                star_count = star_tag['title'].strip()
                return int(star_count.replace(',', ''))
            else:
                print("Star tag not found.")
                return None
        except Exception as e:
            print(f"Error occurred: {e}")
            return None

# 사용 예시
if __name__ == "__main__":
    repo_url = input("Enter the GitHub repository URL: ")
    crawler = GitHubStarCrawler(repo_url)
    star_count = crawler.get_star_count()
    if star_count is not None:
        print(f"The repository has {star_count} stars.")
    else:
        print("Failed to retrieve the star count.")
```
